```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("./divar_car_dataset.csv")
```

```python
df.head(2)
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.head(2)
```

```python
df["price"] = df["price"]/1_000_000
```

```python
df.isnull().sum()
```

```python
sns.scatterplot(df["price"])
```

```python
df = df[(df["price"] >= 200) & (df["price"] <= 40000)]
```

```python
len(df)
```

```python
df.head()
```

```python
df = df.drop(columns=['title'])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(columns=['price'])
y = df['price']
```

```python
x_train , x_test , y_train , y_test = train_test_split(
    x,
    y,
    test_size=0.25 ,
    random_state=42
)
```

```python
import tensorflow as tf
```

```python
x_train.shape
```

```python
Tokenizer = tf.keras.layers.TextVectorization()
```

```python
all_text = ""
```

```python
for col in ["brand" , "gerbox" , "fuel" , "colors" , "motor_status" ,"chassis_status" ,"body_status" ,"gerbox_status"]:
        all_text += df[col].astype(str)
```

```python
all_text = all_text.str.replace('\u200c', ' ')
```

```python
Tokenizer.adapt(all_text)
```

```python
len(Tokenizer.get_vocabulary())
```

```python
vocab_size = len(Tokenizer.get_vocabulary())
embd_len = 16
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(
        input_dim=vocab_size ,
        output_dim=embd_len ,
        mask_zero=True
    ) ,
    tf.keras.layers.LSTM(64, return_sequences=True) ,
    tf.keras.layers.Dropout(0.4) ,
    tf.keras.layers.LSTM(64),
    tf.keras.layers.Dense(1)
])
```

```python
text_cols = ["brand", "gerbox", "fuel", "colors","motor_status", "chassis_status", "body_status", "gerbox_status"]
```

```python
def make_text(d):
    return (d[text_cols].astype(str)
            .agg(" ".join, axis=1)
            .str.replace("\u200c", " ", regex=False))
```

```python
train_text = make_text(x_train)
test_text = make_text(x_test)
```

```python
x_train_tok = Tokenizer(train_text.values).numpy()
```

```python
x_test_tok = Tokenizer(test_text.values).numpy()
```

```python
y_train_log = np.log1p(y_train.values).astype("float32")
```

```python
y_test_log = np.log1p(y_test.values).astype("float32")
```

```python
x_train_tok.shape
```

```python
sns.scatterplot(y_train_log)
```

```python
model.compile(tf.keras.optimizers.AdamW(learning_rate=1e-4, weight_decay=1e-4) , loss="huber" )
```

```python
model.summary()
```

```python
model.fit(
    x_train_tok ,
    y_train_log ,
    epochs=100,
    validation_split=0.1 ,
    batch_size=8
)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
pred = np.expm1(model.predict(x_test_tok).flatten())
real = y_test.values
print(mean_absolute_error(real, pred))
print(r2_score(real, pred))
```

```python
model.summary()
```

```python
model.save("my_model.keras")
```
